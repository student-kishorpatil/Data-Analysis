```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df=pd.read_csv(r"C:\Users\kisho\Downloads\archive (19).zip")
```

```python
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
a=df['Attrition'].value_counts()                                              #Total Employees & Attrition Count
```

```python
a
```

```python
plt.figure(figsize=(6,5))                                                                          #Attrition Visualization

sns.countplot(x='Attrition', data=df)

plt.title("Employee Attrition Count")

plt.show()
```

```python
department_attrition = (df.groupby(['Department', 'Attrition']).size().unstack(fill_value=0))                 #Department-wise Attrition
```

```python
department_attrition
```

```python
department_attrition.plot(kind='bar', figsize=(8,5))

plt.title("Department-wise Attrition")

plt.ylabel("Employees")

plt.show()
```

```python
plt.figure(figsize=(8,5))                                                                                   # Salary Distribution

plt.hist(df['MonthlyIncome'])

plt.title("Monthly Income Distribution")

plt.xlabel("Monthly Income")

plt.ylabel("Employees")

plt.show()
```

```python
overtime_attrition = (df.groupby(['OverTime', 'Attrition']).size().unstack(fill_value=0))                     #Overtime vs Attrition
```

```python
overtime_attrition
```

```python
overtime_attrition.plot(kind='bar')

plt.title("Overtime vs Attrition")

plt.ylabel("Employees")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x='JobSatisfaction',
    hue='Attrition',
    data=df
)

plt.title("Job Satisfaction vs Attrition")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df['Age'])

plt.title("Employee Age Distribution")

plt.xlabel("Age")

plt.ylabel("Employees")

plt.show()
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(df.corr(numeric_only=True))

plt.show()
```
